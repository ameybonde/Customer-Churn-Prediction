```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/ML_Projects/Customer_Churn_Prediction/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.head()
```

```python
df.drop("customerID", axis=1, inplace=True)
```

```python
df.info()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df["Churn"] = df["Churn"].map({"No": 0,"Yes": 1})
```

```python
df.isna().sum().sum()
```

```python
df.dropna(inplace=True)
```

```python
x = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
categorical_features = [col for col in x.columns if x[col].dtype == "O"]
categorical_features
```

```python
numerical_features = [col for col in x.columns if x[col].dtype != "O"]
numerical_features
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
numerical_transformer=StandardScaler()
categorical_transformer=OneHotEncoder(handle_unknown="ignore")


preprocessor=ColumnTransformer(
    [
        ("OneHotEncoder",categorical_transformer,categorical_features),
        ("StandardScaler",numerical_transformer,numerical_features)
    ]
)
```

```python
preprocessor
```

```python
x_train=preprocessor.fit_transform(x_train)
x_test=preprocessor.transform(x_test)
```

```python
x
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import recall_score,accuracy_score,precision_score,f1_score,roc_auc_score
```

```python
# For decision tree
dt_params = {
    "criterion": "gini",
    "max_depth": 5,
    "min_samples_split": 10,
    "min_samples_leaf": 4,
    "class_weight": "balanced"
}
```

```python
# For random forest
rf_params = {
    "n_estimators": 200,
    "max_depth": 10,
    "min_samples_split": 5,
    "min_samples_leaf": 2,
    "max_features": "sqrt",
    "class_weight": "balanced",
    "random_state": 42
}
```

```python
models={
    "Logistic Regression":LogisticRegression(),
    "Decision Tree":DecisionTreeClassifier(**dt_params),
    "Random Forest":RandomForestClassifier(**rf_params)
}

for i in range(len(list(models))):
  model=list(models.values())[i]
  model.fit(x_train,y_train)

  y_train_pred=model.predict(x_train)
  y_test_pred=model.predict(x_test)

  print(model)
  print("Training Accuracy")
  print("Accuracy :", accuracy_score(y_train,y_train_pred))
  print("Precision:", precision_score(y_train,y_train_pred))
  print("Recall   :", recall_score(y_train,y_train_pred))
  print("F1 Score :", f1_score(y_train,y_train_pred))
  print("ROC-AUC  :", roc_auc_score(y_train,y_train_pred))
  print("\n")

  print("Testing Accuracy")
  print("Accuracy :", accuracy_score(y_test,y_test_pred))
  print("Precision:", precision_score(y_test,y_test_pred))
  print("Recall   :", recall_score(y_test,y_test_pred))
  print("F1 Score :", f1_score(y_test,y_test_pred))
  print("ROC-AUC  :", roc_auc_score(y_test,y_test_pred))
  print("-"*20,"\n")
```

## Random forest has highest F!, recall. So we will fine tune it.

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = {
    "n_estimators": [100, 200],
    "max_depth": [5, 10],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "max_features": ["sqrt"],
    "class_weight": ["balanced"]
}
```

```python
grid = GridSearchCV(
    estimator=RandomForestClassifier(**rf_params),
    param_grid=params,
    scoring="recall",
    cv=5,
    verbose=2,
    n_jobs=-1
)
```

```python
grid.fit(x_train, y_train)
```

```python
print("Best Parameters:")
print(grid.best_params_)
```

```python
best_model = grid.best_estimator_
```

```python
y_train_pred = best_model.predict(x_train)
y_test_pred = best_model.predict(x_test)
```

```python
print("Training Metrics")
print("Accuracy :", accuracy_score(y_train, y_train_pred))
print("Precision:", precision_score(y_train, y_train_pred))
print("Recall   :", recall_score(y_train, y_train_pred))
print("F1 Score :", f1_score(y_train, y_train_pred))
print("ROC-AUC  :", roc_auc_score(y_train, y_train_pred))
print("\n")

print("Testing Metrics")
print("Accuracy :", accuracy_score(y_test, y_test_pred))
print("Precision:", precision_score(y_test, y_test_pred))
print("Recall   :", recall_score(y_test, y_test_pred))
print("F1 Score :", f1_score(y_test, y_test_pred))
print("ROC-AUC  :", roc_auc_score(y_test, y_test_pred))
```
